Day 26: How do I find and remove duplicate rows in pandas?

```python
import pandas as pd
```

```python
# read a  dataset of movies reviewers (modifying the default parameter values for read_table)
user_cols = ['user_id', 'age', 'gender', 'occupation', 'zip_code']
users = pd.read_table("https://bit.ly/movieusers", sep="|", header=None, names=user_cols, index_col='user_id')
```

```python
users.head()
```

```python
users.shape
```

Now, if you want to identify duplicate zip codes.
You can use a series method

```python
users.zip_code.duplicated().sum() # Returns true if there is an entry previous to it with the same value, i.e. above it. Using the sum method, the true get converted to 1 and false get converted to 0. This implies that there are 148 duplicate zip codes.
```

For checking duplication in the data frame as a whole instead of series.

```python
users.duplicated().sum() # This will output true, if the entire row, is duplicate or is present before it.
```

In order to see the duplicate rows, we can use the loc method.


```python
users.loc[users.duplicated(keep='first'), : ] # We can pass in series of boolean where the value is true, Mark duplicate as true, except the first occurrence of the row, so here the first occurrence is kept and rest of the remaining once are identified as duplicates.
#users.loc[users.duplicated(keep='last'), : ] 
```

For dropping the duplicates use the following method.

```python
users.drop_duplicates(keep='last').shape # keep='last' and keep='first' return the same result. This will not occur inplace by default.
```

```python
users.drop_duplicates(keep=False).shape # This will drop all the row that are tagged as duplicate i.e. both the entries first 7 and last 7 
```

If you want to consider certain columns when considering duplicates. When identifying duplicates. Example: if you want to keep age+zip code as a unique identifier.

```python
users.duplicated(subset=['age', 'zip_code']).shape# This implies that if you are considering age and zip code, then there are 16 elements.
```

```python
# For dropping the duplicate rows 
users.drop_duplicates(subset=['age', 'zip_code']).shape
```
